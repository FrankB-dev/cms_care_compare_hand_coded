```python
import pandas as pd
```

```python
filepaths = [
    '../data/2019/Complications and Deaths - National.csv',
    '../data/2020/qqw3-t4ie.csv',
    '../data/2021/Complications_and_Deaths-National.csv',
    '../data/2022/Complications_and_Deaths-National.csv',
    '../data/2023/Complications_and_Deaths-National.csv',
    '../data/2024/Complications_and_Deaths-National.csv',
    '../data/2025/Complications_and_Deaths-National.csv',
    '../data/2026/Complications_and_Deaths-National.csv',
]
```

```python
# check columns
col_sets = []
for fp in filepaths:
    df = pd.read_csv(fp)
    cols = set(df.columns.tolist())
    col_sets.append(cols)

all_equal = all(s == col_sets[0] for s in col_sets)
all_equal
```

```python
# check shape
for fp in filepaths:
    print(fp)
    df = pd.read_csv(fp)
    print(df.shape)
# more rows in 2025 and 2026
```

```python
# check measures
measure_id_sets = []

for fp in filepaths:
    print(fp)
    df = pd.read_csv(fp)
    mids = set(df['Measure ID'].tolist())
    measure_id_sets.append(mids)

all_equal = all(s == measure_id_sets[0] for s in measure_id_sets)
all_equal
```
